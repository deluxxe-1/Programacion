### 1. ¿Cuáles son los operadores lógicos?

Los operadores lógicos modifican y unen expresiones evaluadas en contexto booleano para crear condiciones más complejas.

### 3. Operador `or`

```python
c = 5
```

```python
if c is 4 or c is 3:
    print("C es 4 o 3")
else:
    print("C no es 4 ni 3")
```

### 4. Operador `and`

```python
edad = 18
```

```python
carnet = True
```

```python
if edad >= 18 and carnet == True:
    print("Puede conducir")
else:
    print("No puede conducir")
```
